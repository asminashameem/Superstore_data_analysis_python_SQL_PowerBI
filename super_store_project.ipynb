```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.boxplot(x=data['sales'])
```

```python
print("Total Sales:", data['sales'].sum())
print("Total Profit:", data['profit'].sum())
print("Profit Margin %:", data['profit'].sum() / data['sales'].sum() * 100)
```

```python
# Top 5 profitable vs loss-making
data.groupby('sub_category')['profit'].sum().sort_values(ascending=False).head(10)
```

```python
# Sales by Category
data.groupby('category')['sales'].sum().sort_values(ascending=False)
```

```python
# Sales by Region
data.groupby('region')['sales'].sum()
```

```python
# Sales by Segment (Consumer, Corporate, Home Office)
data.groupby('segment')['sales'].sum()
```

```python
# Top 10 Products by Sales
data.groupby('product_name')['sales'].sum().sort_values(ascending=False).head(10)
```

```python
#  Discount vs Profit (imp for project)
plt.scatter(data['discount'], data['profit'])
plt.title('Discount vs Profit - High discount = Loss?')
plt.xlabel('discount')
plt.ylabel('profit')
plt.show()
```

```python
# Sales by Category bar chart
data.groupby('category')['sales'].sum().plot(kind='bar')
plt.title('Sales by Category')
plt.show()
```

```python
# Profit by Region
data.groupby('region')['profit'].sum().plot(kind='bar', color='green')
plt.title('Profit by Region')
plt.show()
```

```python
#  Profit by Sub-Category (to find loss-making products)
data.groupby('sub_category')['profit'].sum().sort_values().plot(kind='barh')
plt.title('Profit by Sub-Category - Tables, Bookcases in loss')
plt.show()
```

```python
#  Sales by Region
data.groupby('region')['sales'].sum().plot(kind='bar', color='orange')
plt.title('Sales by Region')
plt.show()
```

```python
#  top 10 State by states
data.groupby('state')['sales'].sum().sort_values(ascending=False).head(10).plot(kind='barh', color='red')
plt.title('Top 10 States by Sales - California Highest')
plt.gca().invert_yaxis() # California on top
plt.show()
```

```python
data['order_date'] = pd.to_datetime(data['order_date'])
data.groupby(data['order_date'].dt.to_period('M'))['sales'].sum().plot()
plt.title('Monthly Sales Trend - Sales peak in Nov-Dec')
plt.show()
```

```python
# profit by customer segment
data.groupby('segment')['profit'].sum().plot(kind='bar', color='skyblue')
plt.title('Profit by Segment - Consumer Highest')
plt.show()
```

```python
# sales by shipping mode
data.groupby('ship_mode')['sales'].sum().plot(kind='pie', autopct='%1.1f%%')
plt.title('Sales by Ship Mode')
plt.ylabel('')
plt.show()
```
